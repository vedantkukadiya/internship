```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('electronic.csv')
print(df.describe())
df
```

```python
prices = df['Price'].values
Units_Sold = df['Units_Sold'].values

# Statistical Analysis
print("mean of Prices:", np.mean(prices))
print("median of Prices:", np.median(prices))
print("std of Prices:", np.std(prices))

print("mean of Units_Sold:", np.mean(Units_Sold))
print("median of Units_Sold:", np.median(Units_Sold))
print("std of Units_Sold:", np.std(Units_Sold))
```

```python
#visualization
# Histogram for Prices and Units_Sold
plt.hist(prices, bins=20, alpha=0.7, color='blue', label='Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Prices')
plt.legend()
plt.show()

plt.hist(Units_Sold, bins=20, alpha=0.7, color='green', label='Units Sold')
plt.xlabel('Units Sold')
plt.ylabel('Frequency')
plt.title('Distribution of Units Sold')
plt.legend()
plt.show()

# Scatter plot between Prices and Units_Sold
plt.scatter(prices,Units_Sold, alpha=0.4, color='purple', label='Price vs Units Sold')
plt.xlabel('Price')
plt.ylabel('Units Sold')
plt.title('Price vs Units Sold')
plt.legend()
plt.show()

#bar chart for average Units_Sold per Price range
Product = df['Product'].values
plt.bar(Product, Units_Sold, color='orange', alpha=0.6)
plt.xlabel('Product')
plt.ylabel('Units Sold')
plt.title('Units Sold by Product')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Bar chart (category-wise avg price)

Category = df['Category'].values
avg_price_per_category = df.groupby('Category')['Price'].mean()
plt.bar(avg_price_per_category.index, avg_price_per_category.values, color='cyan', alpha=0.6)
plt.xlabel('Category')
plt.ylabel('Average Price')
plt.title('Average Price by Category')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
