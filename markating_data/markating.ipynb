### 1. UNIVARIATE ANALYSIS (One column at a time)

#### A) Numerical Univariate

Non-visual

```python
df.describe()
```

##### 🔸Visual

```python
#histogram

plt.figure(figsize=(10, 6))
plt.hist(df['Income'], bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
#box plot
sns.boxplot(x=df['Income'], color='lightgreen', orient='h', width=0.5, fliersize=5, flierprops=dict(marker='o', markerfacecolor='red', markersize=8, linestyle='none'))
plt.title('Box Plot of Income')
plt.xlabel('Income')
plt.ylabel(' ')
plt.grid(axis='x', alpha=0.75)
plt.show()
```

```python
#kde plot
sns.kdeplot(df['Income'], color='purple', bw_adjust=0.5, alpha=0.7, fill=True, linewidth=2, linestyle='--')
plt.title('Income Density Plot')
plt.xlabel('Income')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#### B) Categorical Univariate

##### 🔸Non-visual

```python
#Marital_Status column analysis

#frequency of test
print('count of each category in Marital_Status column:')
print(df['Marital_Status'].value_counts())

#percentage of test

print('\npercentage of each category in Marital_Status column:')
print(df['Marital_Status'].value_counts(normalize=True) * 100)

#number of unique categories
print('\nnumber of unique categories in Marital_Status column:')
print(df['Marital_Status'].nunique())
```

##### 🔸Visual

```python
#visualization

plt.figure(figsize=(8, 6))
df['Marital_Status'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black', alpha=0.7)
plt.title('')
plt.xlabel('Marital Status')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#pie chart

plt.figure(figsize=(5, 5))
plt.pie(df['Marital_Status'].value_counts(), labels=df['Marital_Status'].value_counts().index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title('Marital Status Distribution')
plt.axis('equal')
plt.show()
```

```python
import seaborn as sns

#countplot
plt.figure(figsize=(8, 6))

sns.countplot(data=df, x='Marital_Status', hue='Marital_Status', palette='Set2', edgecolor='black', alpha=0.7, legend=False)
plt.title('Count Plot of Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### C) Categorical vs Categorical

##### 🔸Non-Visual

```python
#Education vs Responses (Crosstab with percentages)
education_response_ct = pd.crosstab(df['Education'], df['Response'], normalize='index') * 100
print(education_response_ct)
```

```python
#Country vs Responses (Crosstab with percentages)
country_response_ct = pd.crosstab(df['Country'], df['Response'], normalize='index') * 100
print(country_response_ct)
```

```python
#contingency table
contingency_table = pd.crosstab(df['Education'], df['Response'])
print(contingency_table)
```

##### 🔸Visual

```python
#heatmap of Education vs Response contingency table

sns.heatmap(contingency_table, annot=True, fmt='d', cmap='YlGnBu', cbar=True)
plt.title('Heatmap of Education vs Response Contingency Table')
plt.show()
```

```python
#stacked bar plot of Education vs Responses
contingency_table.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis', edgecolor='black', alpha=0.7)
plt.title('Stacked Bar Plot of Education vs Responses')
plt.xlabel('Education')
plt.ylabel('Count')
plt.legend(title='Response', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#countplot of Marital_Status vs Responses
sns.countplot(data=df, x='Marital_Status', hue='Response', palette='Set1', edgecolor='black', alpha=0.7)
plt.title('Count Plot of Marital Status vs Responses')
plt.xlabel('Marital Status')
plt.ylabel('Count')
plt.legend(title='Response')
plt.tight_layout()
plt.show()
```
