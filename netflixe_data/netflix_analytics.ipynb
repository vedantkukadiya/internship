```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path = 'netflixe.csv'
df = pd.read_csv(file_path)
df
```

```python
%pip install tabulate

# --- 2. NumPy/Pandas Descriptive Statistics ---

duration_array = df['Duration'].values 
release_year_array = df['Release_Year'].values

duration_mean = np.mean(duration_array)
duration_median = np.median(duration_array)
duration_min = np.min(duration_array)
duration_max = np.max(duration_array)
duration_std = np.std(duration_array)

duration_mode = df['Duration'].mode().iloc[0] if not df['Duration'].mode().empty else 'N/A'


correlation_matrix = np.corrcoef(release_year_array, duration_array)
correlation_r = correlation_matrix[0, 1]

# 3. Numerical Analysis (Duration & Correlation)
print("\nDuration Statistics:")
duration_stats = pd.DataFrame({
    'Metric': ['Mean', 'Median', 'Mode', 'Min', 'Max', 'Std Dev'],
    'Duration (min/episodes)': [duration_mean, duration_median, duration_mode, duration_min, duration_max, duration_std]
})
print(duration_stats.to_markdown(index=False, numalign="left", stralign="left"))

print(f"\nCorrelation (r) between Release_Year and Duration: {correlation_r:.4f}")
print("Interpretation: A value close to 0 suggests a very weak linear relationship.")
```

```python
# --- 4. Matplotlib Visualization ---
plt.style.use('ggplot')


type_counts = df['Type'].value_counts()
genre_counts_sorted = df['Genre'].value_counts().sort_values(ascending=False)
year_counts = df['Release_Year'].value_counts().sort_index()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: Content Type Distribution
axes[0].bar(type_counts.index, type_counts.values, color=['#E50914', '#221F1F'])
axes[0].set_title('Distribution of Content Type')
axes[0].set_xlabel('Type')
axes[0].set_ylabel('Count')
for i, v in enumerate(type_counts.values):
    axes[0].text(i, v + 0.1, str(v), ha='center', va='bottom')

# Plot 2: Genre Distribution
axes[1].bar(genre_counts_sorted.index, genre_counts_sorted.values, color='#83b429')
axes[1].set_title('Distribution of Genres')
axes[1].set_xlabel('Genre')
axes[1].set_ylabel('Count')
axes[1].tick_params(axis='x', rotation=45)
for i, v in enumerate(genre_counts_sorted.values):
    axes[1].text(i, v + 0.1, str(v), ha='center', va='bottom')

# Plot 3: Content Released by Year
axes[2].bar(year_counts.index.astype(str), year_counts.values, color='#45885c')
axes[2].set_title('Content Released by Year')
axes[2].set_xlabel('Release Year')
axes[2].set_ylabel('Count')
for i, v in enumerate(year_counts.values):
    axes[2].text(i, v + 0.1, str(v), ha='center', va='bottom')

plt.tight_layout()
plt.savefig('netflix_full_analysis_plots.png')
plt.close()

print("Three plots showing the distribution of Content Type, Genre, and Release Year have been saved as 'netflix_full_analysis_plots.png'.")
```
