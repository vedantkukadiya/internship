```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data = np.array([
    [1, 150000, 180000, 200000],
    [2, 160000, 190000, 210000],
    [3, 170000, 200000, 220000],
    [4, 180000, 210000, 230000],
    [5, 190000, 220000, 240000]
])
print("=== zomato salese analisys ===")
```

```python
print("shape of data:",sales_data.shape)
```

```python
print("sample data for 1st 3 restorants",sales_data[0:3])
```

#total salse per year

```python
print(np.sum(sales_data,axis=0))
min_sales = np.min(sales_data[:,1:],axis=0)
max_sales = np.max(sales_data[:,1:],axis=0)
print("min salse :",min_sales)
print("max salse :",max_sales)
avg_sales = np.mean(sales_data[:,1:],axis=0)
print("avg salse :",avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:,1:],axis=1)
print("cumsum :",cumsum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average Sales Over Time")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(true)
plt.show()
```
