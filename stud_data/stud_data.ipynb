```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('stud.csv')
print(df)
df.info()
df.describe()
df.isnull().sum()


gender_avg = df.groupby('gender')[['math_score', 'reading_score', 'writing_score']].mean()

gender_avg.plot(kind='bar', figsize=(8,6), color=['blue','orange','green'], alpha=0.7)
plt.title('Gender-wise average scores')
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.legend(['Math Score', 'Reading Score', 'Writing Score'])
plt.show()
```

```python
#gender wise average score

gender_avg = df.groupby('gender')[['math_score', 'reading_score', 'writing_score']].mean()
print(gender_avg)
gender_avg = df.groupby('gender')[['math_score', 'reading_score', 'writing_score']].median()
print(gender_avg)
```

```python
# Scatter plot for study time vs performance
plt.scatter(df['Study_Time_Hours'], df['reading_score'], color='black', marker='o', alpha=0.5, label='Reading Score')
plt.xlabel('Study Time Hours')
plt.ylabel('Reading Score')
plt.title('Study Time vs Reading Score')
plt.legend()
plt.show()
```
