```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('sales.csv')
print(df)

#tot_revenue = df['Revenue'].sum()
region_revenue = df.groupby('Region')['Revenue'].sum()
print(region_revenue)

#unit wise product sold
pro_units = df.groupby('Product')['Units_Sold'].sum()
print(pro_units)

#catagaroy wise
cat_sales = df.groupby('Category')['Revenue'].sum()
print(cat_sales)

#daliy revenu 
daliy_revenue = df.groupby('Date')['Revenue'].sum()
daliy_revenue
```

```python
#numpy 
unit_sold = np.mean(df['Units_Sold'])
print(unit_sold)

#median unit
unit=np.median(df['Units_Sold'])
print(unit)

#Stander Daviation
std=np.std(df['Units_Sold'])
print(std)

#unit sold by categaroy
price = df['Unit_Price'].to_numpy()

phone_price = df[df['Product'] == 'Phone']['Unit_Price']
laptop_price = df[df['Product']== 'Laptop']['Unit_Price']
hadphone_price=df[df['Product']=='Headphone']['Unit_Price']

print("Phone Price Statistics:",np.mean(phone_price))
print("Laptop Price Statistics:",np.mean(laptop_price))
print("Headphone Price Statistics:",np.mean(hadphone_price))

print("Phone Price Standard Deviation:",np.std(phone_price))
print("Laptop Price Standard Deviation:",np.std(laptop_price))
print("Headphone Price Standard Deviation:",np.std(hadphone_price))

#mathematical compution
total_revenue = np.sum(df["Revenue"].to_numpy())

# Max and Min units sold
max_units = np.max(df["Units_Sold"])
min_units = np.min(df["Units_Sold"])

# Correlation (Units Sold vs Revenue)
correlation = np.corrcoef(df["Units_Sold"], df["Revenue"])[0, 1]

print("Total Revenue:", total_revenue)
print("Max Units Sold:", max_units)
print("Min Units Sold:", min_units)
print("Correlation (Units Sold vs Revenue):", correlation)
```

```python
#line chat using daliy revanue trand

plt.plot(df['Date'],df['Revenue'], color='skyblue',marker='o',linestyle='--',linewidth=2,markersize=6)
plt.xlabel('Date')
plt.ylabel('Revenue')
plt.title('Date vs Revenue')
plt.xticks(rotation=45)
plt.show()

#bar chart product wies revanue
plt.bar(df['Product'],df['Revenue'], color='darkorange',edgecolor='black',align='center',label='Revenue by Product',width=0.5)
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.title('Product vs Revenue')
plt.tight_layout()
plt.legend()
plt.show()

#pie chart region wise revanue
plt.pie(region_revenue, labels=region_revenue.index, autopct='%1.1f%%', startangle=140, colors=['lightcoral', 'lightskyblue', 'lightgreen', 'gold'])
plt.title('Revenue Distribution by Region')
plt.axis('equal')
plt.show()
```
