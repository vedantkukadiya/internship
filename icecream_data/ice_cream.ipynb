```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('ice.csv')
print(df.describe())    
```

```python
# Split the single column into two separate columns
df[['Temperature (°C)', 'Ice Cream Sales (units)']] = df['Temperature (°C)\tIce Cream Sales (units)'].str.split('\t', expand=True)
df['Temperature (°C)'] = df['Temperature (°C)'].astype(float)
df['Ice Cream Sales (units)'] = df['Ice Cream Sales (units)'].astype(float)

# numpy statistics
temp = df['Temperature (°C)'].values
sales = df['Ice Cream Sales (units)'].values

print("mean temperature:", np.mean(temp))
print("median temperature:", np.median(temp))
print("std temperature:", np.std(temp))

print("mean sales:", np.mean(sales))
print("median sales:", np.median(sales))     
print("std sales:", np.std(sales))

print("correlation between temperature and sales:", np.corrcoef(temp, sales)[0, 1])
```

```python
#visualization
#scatter plot
plt.scatter(df['Temperature (°C)'], df['Ice Cream Sales (units)'], color='blue', alpha=0.5, marker='x', label='Data Points')
plt.xlabel('Temperature (°C)')
plt.ylabel('Ice Cream Sales (units)')
plt.title('Ice Cream Sales vs Temperature')
plt.grid()
plt.legend()
plt.show()
```

```python
#line plot
plt.plot(df['Temperature (°C)'], df['Ice Cream Sales (units)'], color='gray', marker='x', linestyle=':', label='Sales Trend')
plt.xlabel('Temperature (°C)')
plt.ylabel('Ice Cream Sales (units)')
plt.title('Ice Cream Sales Trend with Temperature')
plt.grid()
plt.legend()
plt.show()
```

```python
#histogram of sales
plt.hist(df['Ice Cream Sales (units)'], bins=10, color='green', alpha=0.7, edgecolor='black', label='Sales Distribution',width=5, align='mid')
plt.xlabel('Ice Cream Sales (units)')
plt.ylabel('Frequency')
plt.title('Distribution of Ice Cream Sales')
plt.grid()
plt.legend()
plt.show()
```

```python
#histogram of temperature
plt.hist(df['Temperature (°C)'], bins=10, color='purple', alpha=0.7, edgecolor='black', label='Temperature Distribution',width=5, align='mid')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.title('Distribution of Temperature')
plt.grid()
plt.legend()
plt.show()
```
